## 3. My Current Understanding of util.py

`util.py` is not just a file loader.

Its real role is to transform raw, page-level experimental data into a structured user-level analysis representation.

What I currently think it does:
- reads raw records
- groups data by user
- links condition, task behavior, and questionnaire data

## 4. My Current Understanding of calc_measures.py

`calc_measures.py` appears to compute behavior-oriented measures rather than only subjective ratings.

These measures are important because the baseline paper does not only ask whether users trust AI more.
It asks whether users rely on AI appropriately.

The key logic seems to involve:
- user's first decision
- AI recommendation
- user's final decision
- correctness of the final decision

This makes it possible to distinguish:
- general agreement with AI
- switching behavior
- more appropriate vs less appropriate reliance patterns
